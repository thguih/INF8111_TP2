## 2. Construction des transactions

On joint `OBSERVATIONS.csv` et `SYMPTOMS.csv`, puis on regroupe par patient (`VAERS_ID`) avec `COLLECT_SET` pour obtenir la liste des noms de symptômes de chaque patient.

![Figure 4](captures/fig4_transactions.png)

**Figure 4 - Transactions créées à partir de OBSERVATIONS.csv** 
On obtient 2 599 720 transactions (une par patient). L'affichage montre les 5 premières lignes, avec la liste des noms de symptômes. Temps d'exécution : Wall time 10,6 s

## 3. MBA sur le dataset complet

La sortie contient le nom des symptômes et la colonne `patterns` est triée par ordre alphabétique. La colonne `association_rules` est tronquée à l'affichage parce que les listes de règles sont très longues pour les symptômes fréquents. Les patterns de taille 3 ont une liste de règles vide (`[]`), ce qui est normal puisqu'on ne génère pas de règles à partir de patterns de taille 4.
